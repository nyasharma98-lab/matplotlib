```python
import matplotlib.pyplot as plt #matplotlib is python library for creating static, animated & interactive visualizations.
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]
import matplotlib.pyplot as plt
plt.plot(x, y) #line chart
plt.title("plotting")
plt.xlabel("rollno")
plt.ylabel("id")
plt.show()
```

```python
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232] 

import matplotlib.pyplot as plt
plt.plot(years, oscar_revenue)
plt.title("Revenue in each year for oscar movies")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")
```

```python
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232] 

import matplotlib.pyplot as plt
plt.bar(years, oscar_revenue) #bar chart is a graph in which data is represented by rectangular bars.
plt.title("Revenue in each year for oscar movies")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")
```

```python
non_oscar_revenue = [378, 2788, 829, 185, 275]
import matplotlib.pyplot as plt
plt.plot(years, oscar_revenue, label="Oscar Movies",color="red")
plt.plot(years, non_oscar_revenue, label="Non-Oscar Movies")
plt.title("Oscar Movies revenue vs Non - Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.legend()
plt.grid()
```

```python
#a scatter plot shows the relationship b/w two variables by plotting points.
years = [2008, 2009, 2010, 2011, 2012, 2013]
revenue = [1005, 170, 427, 133, 232, 333] 
plt.scatter(years, revenue)
plt.title("Years & Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue")
plt.grid()
```

```python
cities = ["City A", "City B", "City C", "City D", "City E"]
winter_temp = [5, 2, 10, 0, 7]
winter_humidity = [80, 75, 65, 85, 70]

summer_temp = [25, 30, 28, 35, 27]
summer_humidity = [60, 50, 55, 45, 65]

import matplotlib.pyplot as plt
plt.scatter(winter_temp, winter_humidity, label="Winter")
plt.scatter(summer_temp, summer_humidity, label="Summer")
plt.title("Summer vs Winter Data")
plt.xlabel("Temperature")
plt.ylabel("Humidity (%)")
plt.legend()
```
